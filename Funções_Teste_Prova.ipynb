3) Crie uma função com retorno e com parâmetro chamada validar_senha(senha) que recebe uma string e retorna um dicionário contendo:

Chave Significado

tamanho_valido Verdadeiro se tiver ≥ 8 caracteres
tem_numero Verdadeiro se possuir número
tem_maiuscula Verdadeiro se tiver letra maiúscula
tem_minuscula Verdadeiro se tiver letra minúscula
tem_especial Verdadeiro se possuir caractere especial
senha_valida Verdadeiro apenas se todos os critérios forem atendidos

```python
import random
import string
```

```python
def senha():
  senha = ""

  for i in range(2):
    min = random.choice (string.ascii_uppercase)
    mai = random.choice (string.ascii_lowercase)
    num = random.choice (string.digits)
    simb = random.choice (string.punctuation)

    senha += min+mai+num+simb

  return senha
```

```python
print(senha())
```

```python
def validar_senha(senha):
    tamanho =  len(senha) >= 8
    tamanho_valido = any(tamanho for caractere in senha)

    tem_numero = any(caractere.isdigit() for caractere in senha)

    tem_maiuscula = any(caractere.isupper() for caractere in senha)

    tem_minuscula = any(caractere.islower() for caractere in senha)

    caracteres_especiais = "!@#$%^&*()_+-=[]{}|;:,.<>?/"
    tem_especial = any(caractere in caracteres_especiais for caractere in senha)

    senha_valida = all([
        tamanho_valido,
        tem_numero,
        tem_maiuscula,
        tem_minuscula,
        tem_especial
    ])

    return {
        "tamanho_valido": tamanho_valido,
        "tem_numero": tem_numero,
        "tem_maiuscula": tem_maiuscula,
        "tem_minuscula": tem_minuscula,
        "tem_especial": tem_especial,
        "senha_valida": senha_valida
    }
```

```python
validar_senha(senha())
```
