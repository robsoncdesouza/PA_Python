```python
def somatresnotas():
  nota1 = float(input("Digite a primeira nota: "))
  nota2 = float(input("Digite a segunda nota:"))
  nota3 = float(input("Digite a terceita nota:"))
  total = nota1+nota2+nota3
  return total
```

```python
def mediatresnotas():
  soma = somatresnotas()
  media = soma/3
  return media
```

```python
print(f"A média das três notas é {mediatresnotas()}")
```

```python
def calcular_imc(peso, altura):
  imc = peso / (altura**2)
  return imc

peso = float(input("Digite seu peso (kg): "))
altura = float(input("Digite sua altura (m): "))

resultado = calcular_imc(peso, altura)
```

```python
print(f"seu imc é {resultado:.2f}")
```

```python
def calcular_soma_e_media(a,b,c):
  soma = a+b+c
  media = soma/3
  return soma,media

n1 = float(input("digite um número: "))
n2 = float(input("digite um número: "))
n3 = float(input("digite um número: "))

resultado_soma, resultado_media = calcular_soma_e_media(n1,n2,n3)
```

```python
print(f"A soma é {resultado_soma:.2f}")
print(f"A média é {resultado_media:.2f}")
```

```python
def calcular_desconto(valor_original,percentual_desconto):
  percentual_desconto = percentual_desconto/100
  valor_final = valor_original*(1-percentual_desconto)
  return valor_final

preco = float(input("Digite o preço do produto: "))
desconto = float(input("Digite o percentual de desconto: "))
```

```python
print(f"O valor com o desconto fica R${calcular_desconto(preco, desconto)}")
```

```python
def encontrar_extremos(lista):
    if not lista:
        return None, None

    maior = max(lista)
    menor = min(lista)
    return maior, menor

lista = []

while True:
  numero = int(input("Digite um numero e 0 para sair: "))
  if numero == 0:
    break

  lista.append(numero)
```

```python
print(f"O maior e menor número da sua lista são respectivamente: {encontrar_extremos(lista)}")
```

```python
def calculadora(a, b, operacao):
    operacao = operacao.lower().strip()

    if operacao == 'soma':
        return a + b
    elif operacao == 'subtração':
        return a - b
    elif operacao == 'multiplicação':
        return a * b
    elif operacao == 'divisão':
        if b == 0:
            return "Erro: Divisão por zero não é permitida"
        return a / b
    else:
        return "Erro: Operação inválida. Utilize: 'soma', 'subtração', 'multiplicação' ou 'divisão'"

a = int(input("Digite um número: "))
b = int(input("Digite mais um número: "))
operacao = str(input("Digite a operação: "))
```

```python
print(f"O resultado do cálculo é {calculadora(a,b,operacao)}")
```

```python
def enviar_email(destinatario, assunto, corpo="Este é um e-mail automático."):
    print(f"------- E-mail enviado -------")
    print(f"Para: {destinatario}")
    print(f"Assunto: {assunto}")
    print(f"Corpo: {corpo}")
    print("-" * 20)

email = input("Digite o nome do destinatário: ")
assunto = input("Digite o assunto do e-mail: ")
```

```python
def avaliar_aluno(nota):
    if nota >= 7:
        return "Aluno Aprovado"
    elif nota >= 5 and nota < 7:
        return "Aluno de Recuperação"
    else:
        return "Aluno Reprovado"

nota = float(input("Digite a nota do aluno: "))
```
