#Constantes


##Calcule a Área do Círculo

```python
#entrada
raio = float(input("Digite o Valor do Raio: "))
```

```python
#processamento
areacirculo = 3.14159 * (raio **2)
```

```python
#saída
print(f"A área do Círculo é: {areacirculo}")
```

#Calcule a Área do Quadrado

```python
#entrada
lado = float (input ("Digite o lado do Quadrado: "))
```

```python
#processamento
areaquadrado = lado*2
```

```python
#saída
print (f"A Área do Retângulo é: {areaquadrado} ")
```

##Calcule a Área do Retângulo


```python
#entrada
base = float (input ("Digite a base do Retângulo: "))
altura = float(input ("Digite a altura do Retângulo: "))
```

```python
#processamento
arearetangulo = base * altura
```

```python
#saída
print (f"A Área do Retângulo é: {arearetangulo} ")
```

##Perímetro do Círculo

```python
#entrada
raio = float (input ("Digite o valor do Raio"))
PI = 3.1416
```

```python
#processamento
perimetrocirculo = 2 * PI * raio
```

```python
#sída
print (f"O Perímetro do Círculo é: {perimetrocirculo}")
```

#Conversão de Temperatura

```python
#entrada
fahrenheit = float (input ("O valor do Fahrenheit é: "))
```

```python
#processamento
celcius = (fahrenheit-32) * 5/9
```

```python
#sída
print (f"A temperatura em Celsius é: {celcius:.2f}")
```

#Cálculo do IMC

```python
#entrada
peso = float (input ("Seu peso é: "))
altura = float (input ("Sua altura é: "))
```

```python
#processamento
imc = (peso / altura **2)
```

```python
#saída
print (f"Seu Índice de Massa Corporal é: {imc:.2f}")
```
