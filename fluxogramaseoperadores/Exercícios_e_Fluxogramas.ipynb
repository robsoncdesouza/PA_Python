##Cálculo de Média

O fluxograma do cálculo de média pode ser convertido em código Python utilizando os operadores de adição (+) e divisão (/).

image.png

```python
#entrada
nota1 = float(input('Digite o valor da nota 1: '))
nota2 = float(input('Digite o valor da nota 1: '))
nota3 = float(input('Digite o valor da nota 1: '))

#processamento
média = (nota1+nota2+nota3)/3

#saída
print ('A média das notas é:', média)
```

image.png

##Cálculo de Comissão

Outro exemplo é o cálculo de comissão. O fluxograma é convertido em código Python, utilizando o preço unitário, a quantidade vendida e o cálculo da comissão.

image.png

```python
#entrada
precounitario = float(input('Digite o preço unitário da peça: '))
quantidadevendida = float(input('Digite a quantidade vendida: '))

#processamento
valortotalvenda = precounitario * quantidadevendida
comissao = valortotalvenda * 5/100

#saída
print ('O valor total da venda é:', valortotalvenda)
print ('O valor total da comissão é:', comissao)
```

##Custo ao Consumidor de um Carro Novo

O custo ao consumidor de um carro novo é a soma do custo de fábrica com a percentagem do distribuidor e dos impostos (aplicados, primeiro os impostos sobre o custo de fábrica, e depois a percentagem do distribuidor sobre o resultado). Supondo que a percentagem do distribuidor seja de 28% e os impostos 45%. Escrever um algoritmo que leia o custo de fábrica de um carro e informe o custo ao consumidor do mesmo.

image.png

```python
#entrada
custofabrica = float(input('Digite o custo de fábrica'))
imposto = 45
distribuicao = 28

#processamento
imposto = (custofabrica * 45/100) + custofabrica
distribuicao = (imposto * 28/100) + imposto

#saída
print ('O custo ao consumidor é:', distribuicao)
```

##Calculo de hora extra

Um fluxograma pode representar o processo de cálculo de horas a mais a ser recebida, mostrando as etapas de entrada das variáveis, a operação a ser feita com divisão e soma para obter o valor final.

image.png

```python
#entrada
trabalhohr = float(input('Digite as horas de trabalho: '))
trabalhomin = float(input('Digite os minutos de trabalho: '))
ganhohr = float(input('Digite quanto você ganha por hora: '))

#processamento
conversaomin = trabalhomin/60+trabalhohr
salarioextra = (conversaomin*ganhohr) *70/100+conversaomin

#saída
print(f'O valor do salário extra é: {salarioextra} ')
```

##Simulação de Investimento em CDB com Juros Compostos

Uma pessoa deseja investir um valor em um CDB que rende uma taxa fixa de X% ao ano. Escreva um algoritmo que solicite o valor inicial do investimento, a taxa de juros anual (em %) e o período em anos. Depois, calcule o montante final considerando juros compostos.

image.png

```python
#entrada
valorinvestimento = float(input('Digite o valor do investimento:'))
taxajurosano = float(input('Digite a taxa de juros anual:'))
anos = float(input('Digite o tempo de investimento em anos:'))

#processamento
conversaoporcent = taxajurosano/100
valordepoisdotempo = valorinvestimento * (1 + conversaoporcent)  ** anos

#saída
print(f'O valor após o investimento será: {valordepoisdotempo :.2f}')
```

##Divisão de Conta em um Restaurante com Gorjeta

Um grupo de amigos foi a um restaurante e deseja dividir a conta igualmente entre eles. Além disso, o restaurante cobra uma taxa de serviço de 10% sobre o valor total da conta mais uma gorjeta.

Escreva um algoritmo que solicite o valor da conta e o número de pessoas no grupo. Depois, calcule o valor total com a taxa de serviço e quanto cada pessoa deve pagar.

image.png

```python
#entrada
valorconta = float (input('Digite o valor da conta:'))
numeropessoasgp = float (input('Digite o número de pessoas por grupo:'))
valorgorjeta = float (input('Digite o valor da gorjeta:'))
taxa = 10/100
#processamento
taxaservico = taxa * valorconta
valortotal = (taxaservico + valorconta + valorgorjeta) /numeropessoasgp

#saída
print(f'O valor a ser pago é: {valortotal :.2f}')
```
